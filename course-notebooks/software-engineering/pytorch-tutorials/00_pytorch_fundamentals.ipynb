```python
import torch
```

```python
# Scalar
scalar = torch.tensor(7)
print(f"quantity of dimensions: {scalar.ndim}")
print(f"quantity of items: {scalar.item()}")

print("gdgd")
```

```python
# Vector
vector = torch.tensor([7, 7])
print(f"quantity of dimensions: {vector.ndim}")
print(f"shape: {vector.shape}")
```

```python
# Matrix
MATRIX = torch.tensor([[7, 8, 7],
                       [9, 10, 7],
                       [11, 12, 7]])
print(f"quantity of dimensions: {MATRIX.ndim}")
print(f"shape: {MATRIX.shape}")
```

```python
# Tensor
TENSOR = torch.tensor([[[1, 2, 3],
                       [3, 6, 9],
                       [2, 4, 5]],
                       MATRIX])
print(f"quantity of dimensions: {TENSOR.ndim}")
print(f"shape: {TENSOR.shape}")
```

```python
TENSOR2 = torch.tensor([[[0]]])
TENSOR2.shape
```

```python
# Create a random tensor of size (3, 4)
random_tensor = torch.rand(size=(3, 3, 4))
random_tensor, random_tensor.dtype, random_tensor.shape
```

```python
# Create a random tensor of size (224, 224, 3)
random_image_size_tensor = torch.rand(size=(224, 224, 3))
print(f"quantity of dimensions: {random_image_size_tensor.ndim}")
print(f"shape: {random_image_size_tensor.shape}")
random_image_size_tensor
```

```python
zeros = torch.zeros(size=(3, 4))
```

```python
# Create a tensor of all ones
ones = torch.ones(size=(3, 4))
ones, ones.dtype, ones.shape
```

```python
# Create a range of values from 0 to 10
zero_to_ten = torch.arange(start=0, end=10, step=1)
zero_to_ten, zero_to_ten.shape, zero_to_ten.dtype
```

```python
# create a tensor filled with zeros with the size as the input tensor
ten_zeros = torch.zeros_like(input=zero_to_ten)
ten_zeros
```

```python
# Tensor datatypes
float_32_tensor = torch.tensor(
    [3.0, 6.0, 9.0],
    dtype=torch.float64,
    device=None,
    requires_grad=False
    )

float_32_tensor.dtype
```

```python
# Getting information from tensors

## Create a tensor
some_tensor = torch.rand(3, 4)

## Get details of the tensor
print(some_tensor)
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Device of tensor: {some_tensor.device}")
```
