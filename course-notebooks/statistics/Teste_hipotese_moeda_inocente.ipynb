a) Suponha que nos 10 lançamentos, observamos \(Y = 10\) caras. Pode-se rejeitar \(H_{0}\) em favor de \(H_{1}\).

```python
from scipy.stats import binom

n = 10
p = 0.5

# P(Y = 10)
prob_10 = binom.pmf(10, n, p)

# p-valor para Y = 10
p_valor_10 = prob_10 + binom.pmf(0, n, p)

print(f"P(Y = 10): {prob_10:.6f}")
print(f"p-valor para Y = 10: {p_valor_10:.6f}")

alpha = 0.05
if p_valor_10 < alpha:
    print("Rejeitamos H0: a moeda não é justa.")
else:
    print("Não rejeitamos H0: a moeda pode ser justa.")
```

## Análise

Temos como hipótese nula $H_0: p = 0.5$ e hipótese alternativa $H_1: p \neq 0.5$, ou seja, a hipótese nula é de que a moeda é justa, enquanto a hipótese alternativa indica que a moeda é viciada.

$$P(Y=10) = \binom{10}{10} (0.5)^{10} (1-0.5)^{10-10} = 0.000977$$

Como o p-valor de $0.001953$ (aproximadamente $0.20\%$) é menor que o nível de significância padrão de $5\%$, nós rejeitamos $H_0$ em favor de $H_1$. Portanto, temos fortes evidências estatísticas de que a moeda não é justa.

b) Suponha que \(Y = 7\), o que pode-se concluir nesse caso?

```python


# 1. Calcula a probabilidade acumulada até 3 caras (Cauda Inferior)
cauda_inferior = binom.cdf(3, n, p)
print(f"Valor isolado de binom.cdf(3, n, p): {cauda_inferior:.6f}")

# 2. Multiplica por 2 para obter o p-valor bicaudal completo (Cauda Inferior + Cauda Superior)
p_valor_7 = cauda_inferior * 2
print(f"p-valor final bicaudal para Y=7: {p_valor_7:.6f} ({p_valor_7 * 100:.2f}%)")

if p_valor_7 < alpha:
    print("Rejeitamos H0: a moeda não é justa.")
else:
    print("Não rejeitamos H0: a moeda pode ser justa.")
```

## Análise

Como o p-valor bicaudal para $Y = 7$ é de $34.38\%$, ele é significativamente maior do que o nível de significância padrão de $5\%$.

Portanto, **não podemos rejeitar a hipótese nula** $H_0$ de que a moeda é justa. Isso significa que a ocorrência de 7 caras em 10 lançamentos é um resultado estatisticamente comum para uma moeda equilibrada, não havendo evidências suficientes para afirmar que ela é viciada.
