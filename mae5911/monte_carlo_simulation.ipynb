```r
# Z ~ Ber(theta)
# z ~ 0.3
theta = 0.3
n = 1000
MC = 10000
A = numeric(MC)

for(i in 1:MC) {
  z = rbinom(n, size = 1, prob = theta)
  EZ = theta
  VZ = theta * (1 - theta)
  t = sqrt(n) * (mean(z) - EZ)/sqrt(VZ)
  A[i] = t
}

f = function(x) dnorm(x)
hist(A, prob=TRUE)
curve(f, add=TRUE, col='tomato', lwd=2)
```

```r
# Z ~ Poisson(theta)
# z ~ 0.3
theta = 1
n = 100
MC = 10000
A = numeric(MC)

for(i in 1:MC) {
  z = rpois(n, lambda = theta)
  EZ = theta
  VZ = theta
  t = sqrt(n) * (mean(z) - EZ)/sqrt(VZ)
  A[i] = t
}

f = function(x) dnorm(x)
hist(A, prob=TRUE)
curve(f, add=TRUE, col='tomato', lwd=2)
```
