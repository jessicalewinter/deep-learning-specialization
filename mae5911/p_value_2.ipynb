```r
#Z ~ N_2(mu, I), mu = (mu1, mu2)
# H0: mu1 = mu2 =0
# H0' mu1 = mu2

n = 10
mu1 = 10
mu2 = 10
A = list()
k = 0
MC=10000

for (i in 1:MC) {
  z = matrix(rnorm(2*n),n,2)

  T1 = n * sum(colMeans(z)^2)
  T2 = n * diff(colMeans(z))^2
  valor.p1 = pchisq(T1, 2)
  valor.p2 = pchisq(T2, 1)

  # espero que valor.p1 < valor.p2

  if(valor.p1 > valor.p2 & valor.p1 > 0.1 & valor.p2 < 0.05) {
      k=k+1
      A[[k]] = list()
      A[[k]]$sample = z
      A[[k]]$valorp1 = valor.p1
      A[[k]]$valorp2 = valor.p2
  }
}

A[[1]]


```
