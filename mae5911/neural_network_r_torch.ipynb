```r
# Install the torch package from CRAN
install.packages("torch")

# Load the library
library(torch)

require('torch')
```

```r
n = 1000
x = runif(n, -10, 10)
y = f(x)+rnorm(n)*0.3


myModel <- torch::nn_module(
  initialize = function(in_dim, out_dim, N) {
    self$seq <- torch::nn_sequential (
      torch::nn_linear(in_dim, round(N)),
      torch::nn_gelu(),
      torch::nn_linear(round(N), out_dim),
    )
  })
  forward = function(x) {
    mu <- self$seq(x)
    return (mu)
  }

  model = myModel(1,1,3)
  model$parameters
  a=0; for(i in 1:length((model$parameters))) a = a

  yy = torch_tensor(y)$unsqueeze(-1)
  xx = torch_tensor(x)$unsqueeze(-1)
  model(xx)

  optim = optim_adamw(model$parameters, lr = 0.01)
  epochs = 1000
  i=0
  loss = numeric()

  while(i < epochs) {
    i = i+1
    yhat = model(xx)
    L = (1/n) * torch_sum((yy-yhat)^2) # funcao de estimacao
    loss[i] = L$item()
    optmin$zero_grad()
    L$backward() # derivada
    model$parameters[[1]]
    optmin$step()

    if(i%%10==0) {
      cli::cli_progress_message(paste("Época: ",  ))
      par(mfrow = x(1,2))
      plot(y ~ x)
      points(as.matrix(yhat) ~x, col="tomate")
      plot(loss, xlab= "Épocas", ylab="Perda")
    }
  }
```
